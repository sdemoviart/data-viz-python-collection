```python
import pandas as pd
import matplotlib.pyplot as plt
from cycler import cycler
```

# Gráfico de Pizza Simples - plt.pie()

```python
categorias = ['Eletrônicos', 'Roupas', 'Alimentos', 'Livros']
quantidades = [45, 30, 15, 10] # valores em porcentagem


plt.pie(quantidades, labels=categorias, autopct='%1.2f%%')
plt.title('Distribuição % de categorias de produtos')
plt.show()
```

# Gráfico de Pizza com Área em Destaque - plt.pie(explode=)

Para criar uma área em destaque, devemos utilizar o parâmetro explode=argumento(variável com os destaques explícitos). No exemplo abaixo, iremos destacar a área de maior valor (45%)

```python
categorias = ['Eletrônicos', 'Roupas', 'Alimentos', 'Livros']
quantidades = [45, 30, 15, 10] # valores em porcentagem

destaques = [0.1, 0, 0, 0]
cores = plt.get_cmap('Pastel1').colors #aproveitamos e mudamos a paleta de cores.

plt.pie(quantidades, labels=categorias, autopct='%1.2f%%', explode=destaques, colors=cores)
plt.title('Distribuição % de categorias de produtos')
plt.show()
```

```python
import matplotlib.pyplot as plt

# listas com os dados
fabricante = ['Samsung', 'Apple', 'Huawei', 'Xiaomi', 'Oppo', 'Outros']
participacao = [30.2, 26.6, 10.4, 9.7, 4.8, 18.3]


# Config do set de cores
cores = plt.get_cmap('Set2').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle=ciclo_cores)

# Criação dos destaques
destaques = [0.1, 0.05, 0, 0, 0, 0]

# Gráfico
plt.pie(participacao, labels=fabricante, autopct='%.1f', explode=destaques)
plt.title('Participação Percentual de Mercado de Fabricantes de Smartphones')
plt.show()
```
