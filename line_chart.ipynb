# Libraries

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from cycler import cycler
```

# Line Chart - Most Easye

```python

meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']

vendas = [85, 90, 78, 92, 88, 96, 94, 98, 100, 102, 105, 108]


plt.plot(meses, vendas);
plt.xlabel('Mês');
plt.ylabel('Vendas em milhares');
plt.title('Crescimento  das Vendas ao Longo do Ano');
plt.show();
```

# Line Chart - marker and grid lines.
The `plot` function accepts the parameter `'marker'` as an argument, where we indicate the marker format. In the example below, we pass the character `o` to the parameter.

We also have the `.grid` function, which adds grid lines to the chart.

### Tradução
A função `plot` aceita como como argumento o parâmetro `'marker'`, onde devemos indicar o formato do marcador. No exemplo a abaixo, passamos para o parâmetro o caractere 'o'.
Também temos a função `.grid`, que coloca linhas de grade no gráfico.

```python
meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
vendas = [85, 90, 78, 92, 88, 96, 94, 98, 100, 102, 105, 108]

plt.plot(meses, vendas, marker='o')
plt.xlabel('Mês')
plt.ylabel('Vendas (em milhares)')
plt.title('Crescimento das Vendas ao Longo do Ano')
plt.grid(True)
plt.show()
```

# Line Chart - Annotate
The `.annotate` function is used to add annotations (labels, explanatory text) at specific points on a chart. It accepts a few parameters.


### Tradução
A função .annotate  é usada para adicionar anotações (rótulos, textos explicativos) em pontos específicos de um gráfico. Aceita alguns parâmetros.

```python
plt.plot(meses, vendas)

plt.annotate('Menor Valor', color='red', xy=(2,78), xytext=(3, 80), arrowprops=dict(facecolor='red'))

plt.xlabel('Mês')
plt.ylabel('Vendas em milhares')
plt.title('Crescimento  das Vendas ao Longo do Ano')
plt.show()
```

# Line Chart - Annotate and Axis Vertical Span (axvspan)
The `axvspan` is one of Matplotlib’s most powerful tools for data storytelling. While most functions are used to plot the data itself, `axvspan` helps provide context for that data.

Basically, it draws a vertical rectangle that spans the full height of the chart, highlighting a specific interval on the X-axis.


### Tradução
A função axvspan (abreviação de Axis Vertical Span) é uma das ferramentas mais poderosas do Matplotlib para o Data Storytelling. Enquanto a maioria das funções serve para plotar os dados em si, o axvspan serve para contextualizar esses dados.

Basicamente, ela desenha um retângulo vertical que preenche toda a altura do gráfico, destacando um intervalo específico no eixo X.

```python

#alterando a colormap padrão para 'Pastel2'
cores = plt.get_cmap('Pastel2').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle=ciclo_cores)

marcadores = np.where(vendas == np.min(vendas), 'o', '')

plt.plot(meses, vendas)

for i, marcador in enumerate(marcadores):
    if marcador == 'o':
        cor = 'red'
    else:
        cor = 'blue'
    plt.scatter(meses[i], vendas[i], marker=marcador, color=cor)

plt.annotate('Menor Valor', color='red', xy=(2,78), xytext=(2.2, 78))


#destaque da região - axvspan = eixo x vertical espandido
# parametro alpha = transparencia no matplotlib que vai de 0 a 1
plt.axvspan(6, 11, alpha=0.3, color=cores[1])

plt.text(6,85, 'Recuperação de Crescimento', color = 'green', fontsize=9, weight='bold')



plt.xlabel('Mês')
plt.ylabel('Vendas em milhares')
plt.title('Crescimento  das Vendas ao Longo do Ano')
plt.show()
```
