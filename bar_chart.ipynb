# Bar Chart - Most Easye

A bar plot maps a categorical variable (names, groups, types) to a numerical variable. The heigh (or length) of the bar represents its value

### Tradução
O gráfico de barras mapeia uma variável categórica (nomes, grupos, tipos) a uma variável numérica. A altura (ou comprimento) da barra representa o valor.


```python
cidades = ['São Paulo', 'Rio de Janeiro', 'Salvador']
temperaturas = [20, 25, 27]


plt.bar(cidades, temperaturas)
plt.xlabel('Cidades')
plt.ylabel('Temperaturas (em graus Celsius)')
plt.title('Temperaturas em Cidades Brasileiras')
plt.show()
```

# Bar Label - plt.bar_label()
The ax.bar_label() function is the most efficient and “clean” way to add data labels to bar charts. The function identifies where the bars are and positions the text perfectly, whether at the top or in the center of the bar.

The key to using the function is to capture the bar container and apply it directly.

A powerful parameter of the function is `fmt`, which formats the bar values.

### Tradução
A função ax.bar_label() é a forma mais eficiente e "limpa" de adicionar rótulos de dados em gráficos de barras. A função identifica onde as barras estão e posiciona o texto perfeitamente, seja no topo ou no centro da barra.

O segredo da função está em capturar o "container" das barras e aplicar diretamente.

Um parâmetro forte da função é `fmt` que realiza a formatação dos valores das barras.



```python
cidades = ['São Paulo', 'Rio de Janeiro', 'Salvador']
temperaturas = [20, 25, 27]

barras = plt.bar(cidades, temperaturas, color='green') #colocamos o gráfico em um objeto/variável/container, para poder utilizar a função bar_label.

plt.bar_label(barras, labels=temperaturas, padding=3, fontsize=10, fontweight='bold')

plt.xlabel('Cidades')
plt.ylabel('Temperatura Média (°C)')
plt.title('Comparação da Temperatura Média entre Cidades')
plt.show()
```

```python
ano = [2009, 2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022]
receita = [42.9, 65.2, 108.2, 156.5, 170.9, 182.8, 233.7, 215.6, 229.2, 265.6, 260.2, 274.5, 365.8, 394.3]

plt.figure(figsize=(10,4))

barras = plt.bar(ano, receita)
plt.bar_label(barras, fmt='$ %.2f', fontsize=8, padding=2) # função específica de barras.
plt.xticks(ano, rotation=45)


plt.xlabel('Ano')
plt.ylabel('Receita (em bilhões de dólares)')
plt.title('Evolução anual da receita da Apple')
plt.show()
print(barras)
```

## Bar Plot stacked



O Stacked Bar Plot (Gráfico de Barras Empilhadas) é uma evolução do gráfico de barras comum, projetado para mostrar não apenas o total de uma categoria, mas também a sua composição interna.

Para Storytelling, é a ferramenta ideal quando queremos responder a duas perguntas simultaneamente.
Diferente das barras lado a lado (agrupadas), aqui as subcategorias são colocadas umas sobre as outras. A altura total da barra representa o acumulo, enquanto cada segmento colorido representa a parte de um todo.

Para que a função funcione, precisaremos utilizar o parâmetro `bottom`. Plotamos a primeira categoria normalmente e, para a segunda, informamos que ela deve começar onde a primeira terminou.

```python
import pandas as pd
import matplotlib.pyplot as plt

anos = [2018, 2019, 2020, 2021]
vendas_online = [5, 10, 15, 20]
vendas_loja_fisica = [10, 8, 6, 4]

plt.bar(anos, vendas_online, label='Online')
plt.bar(anos, vendas_loja_fisica, bottom=vendas_online, label='Loja Física')

plt.xticks(ticks=anos)
plt.xlabel('Ano')
plt.ylabel('Vendas (em milhões de reais)')
plt.title('Vendas Online vs Lojas')
plt.legend()

plt.show()
```

```python
# Otimizando e melhorando exemplo anterior
anos = [2018, 2019, 2020, 2021]
vendas_online = [5, 10, 15, 20]
vendas_loja_fisica = [10, 8, 6, 4]

# atribuir as barras a variáveis, possibilitando pegar os conjuntos de barras e mudar as propriedades
barras_online = plt.bar(anos, vendas_online, label='Online')
barras_loja_fisica = plt.bar(anos, vendas_loja_fisica, bottom=vendas_online, label='Loja Física')

#Declarando as duas barras
plt.bar_label(barras_online, labels=vendas_online, label_type='center')
plt.bar_label(barras_loja_fisica, labels=vendas_loja_fisica, label_type='center')


# Força o matplotlib a exibir os números de anos como inteiros, se comentar essa linha do código, você vai perceber que ficou estranho.
plt.xticks(ticks=anos)

plt.box(False)
plt.tick_params(axis='x', length=0)
plt.xlabel('Ano')
plt.yticks([])
#plt.ylabel('Vendas (em milhões de reais)')
plt.title('Vendas (em milhões) Online vs Lojas')
plt.legend()

plt.show()
```

# Horizontal Bar Chart - plt.barh()
The Horizontal Bar Chart, implemented using the plt.barh() function, is often the underrated hero of data storytelling. While the vertical bar chart is the standard, the horizontal version solves design problems that the vertical version simply can’t handle well.

The main technical difference is that the axes are reversed:
* In plt.bar(), the first argument is the category (X-axis) and the second is the value (Y-axis).
* In plt.barh(), the first argument is the vertical position (Y-axis) and the second is the bar width (X-axis).

### Tradução
O Gráfico de Barras Horizontais, implementado pela função plt.barh(), é muitas vezes o herói subestimado do Data Storytelling. Enquanto o gráfico de barras vertical é o padrão, o horizontal resolve problemas de design que a versão vertical simplesmente não consegue lidar bem.

A principal diferença técnica é que os eixos são invertidos:
* No plt.bar(), o primeiro argumento é a categoria (eixo X) e o segundo é o valor (eixo Y).
* No plt.barh(), o primeiro argumento é a posição vertical (eixo Y) e o segundo é a largura da barra (eixo X).

```python
produtos = ['Prouduto A', 'Produto B', 'Produto C', 'Produto D', 'Produto E']
quantidades = [23, 45, 17, 30, 37]

barras = plt.barh(produtos, quantidades)
plt.bar_label(barras, labels=quantidades, padding=3, fontsize=10, fontweight='bold')
plt.box(False)

plt.xticks([])
plt.tick_params(axis='y', length=0)
plt.title('Quantidade de Produtos Vendidos')
plt.show()
```

```python
# Destacando a Barra Horizontal de maior valor:
produtos = ['Prouduto A', 'Produto B', 'Produto C', 'Produto D', 'Produto E']
quantidades = [23, 45, 17, 30, 37]
#==========================================================================================

# list comprehesion para remover os rótulos das barras.
rotulos = [valor if valor == max(quantidades) else '' for valor in quantidades]
#==========================================================================================

#definindo padrões de cores diferentes das do padrão do matplotlib
cores = plt.get_cmap('Pastel2').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle=ciclo_cores)

#==========================================================================================
barras = plt.barh(produtos, quantidades)
plt.bar_label(barras, labels=rotulos, padding=3, fontsize=10, fontweight='bold')
plt.box(False)

#==========================================================================================
plt.xticks([])
plt.tick_params(axis='y', length=0)

#==========================================================================================
# Alterando cores das demais barras se não forem o valor máximo
for i, barra in enumerate (barras):
    if i != quantidades.index(max(quantidades)):
        barras[i].set_color(cores[1])
        barras[i].set_alpha(0.3)

#==========================================================================================
plt.title('Quantidade de Produtos Vendidos')
plt.show()
```

# Ordenando Barras e Tornando o Código mais Limpo:


## Entendendo variáveis e organizando dados em ordem decrescente.

```python
#==========================================================================================
# entendo as variáveis 'quantidades' e 'produtos'
print(quantidades)
print(produtos)

print('==========================================================================================')

#alterando uma lista com base em outra
#juntar duas listas no mesmo objeto
print(zip(quantidades, produtos)) #objeto que existe apenas na memória

print('==========================================================================================')

# Gerando uma lista de tuplas com: quantidade e produto, respectivamente.
print(list(zip(quantidades, produtos)))

print('==========================================================================================')

# Ordenando as tuplas por quantidades: o método sorted, automaticamente gera uma lista, logo, não precisamos passar list para transformar em uma lista.
# A partir disso, poderemos começar a ordenar nossas barras
print(sorted(zip(quantidades, produtos)))

print('==========================================================================================')

dados_ordenados = sorted(zip(quantidades, produtos)) #atribuindo método anterior a uma variável (dados_ordenados)
print(dados_ordenados)

print('==========================================================================================')
quantidades_ordenadas = [dado[0] for dado in dados_ordenados] #list comprehension
print(quantidades_ordenadas)

print('==========================================================================================')
produtos_ordenados = [dado[1] for dado in dados_ordenados]
print(produtos_ordenados)

print('==========================================================================================')
```

```python
# Refazendo o código com as novas variáveis (quantidades_ordenadas e produtos_ordenados)

#==========================================================================================
# list comprehension e removendo os rótulos das barras com exceção do valor máximo.
rotulos = [valor if valor == max(quantidades_ordenadas) else '' for valor in quantidades_ordenadas]
#==========================================================================================

# definindo padrões de cores em tempo de execução
cores = plt.get_cmap('Pastel2').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle=ciclo_cores)
#==========================================================================================

#criando objeto barh.
barras = plt.barh(produtos_ordenados, quantidades_ordenadas)
plt.bar_label(barras, labels=rotulos, padding=3, fontsize=10, fontweight='bold')
plt.box(False)
#==========================================================================================

#removendo ticks do eixo x e y
plt.xticks([])
plt.tick_params(axis='y', length=0)

#==========================================================================================
# Alterando cores das demais barras se não forem o valor máximo
for i, barra in enumerate (barras):
    if i != quantidades_ordenadas.index(max(quantidades_ordenadas)):
        barras[i].set_color(cores[1])
        barras[i].set_alpha(0.3)

#==========================================================================================

plt.title('Quantidade de Produtos Vendidos')
plt.show()
```

## Invertendo a Ordem das Barras
É possível inverter a ordem do eixo y para que o maior valor fique na base, através da função plt.gca() -> plt.gca().invert_yaxis()

```python
#==========================================================================================
#Repetindo o código anterior
#==========================================================================================

rotulos = [valor if valor == max(quantidades_ordenadas) else '' for valor in quantidades_ordenadas]


cores = plt.get_cmap('Pastel2').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle=ciclo_cores)

barras = plt.barh(produtos_ordenados, quantidades_ordenadas)
plt.bar_label(barras, labels=rotulos, padding=3, fontsize=10, fontweight='bold')
plt.box(False)

plt.xticks([])
plt.tick_params(axis='y', length=0)


for i, barra in enumerate (barras):
    if i != quantidades_ordenadas.index(max(quantidades_ordenadas)):
        barras[i].set_color(cores[1])
        barras[i].set_alpha(0.3)


plt.title('Quantidade de Produtos Vendidos')
#==========================================================================================

#aplicando função
plt.gca().invert_yaxis()
plt.show()
```

## Voltando para barras verticais após todo o tratamento horizontal
E se, por algum motivo, desejássemos voltar para barras com eixos verticais e não mais horizontais?



```python
#==========================================================================================
rotulos = [valor if valor == max(quantidades_ordenadas) else '' for valor in quantidades_ordenadas]
#==========================================================================================

cores = plt.get_cmap('Pastel2').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle=ciclo_cores)

#==========================================================================================
# 1. mudando o plt.barh para plt.bar
barras = plt.bar(produtos_ordenados, quantidades_ordenadas)
plt.bar_label(barras, labels=rotulos, padding=3, fontsize=10, fontweight='bold')
plt.box(False)

#==========================================================================================
# 2. Está no final do código, deve ser alterado lá.
# 3. Trocar a lista vazia do eixo x pela do eixo y, alterando a função
plt.yticks([])

#==========================================================================================
# 4. Trocar os parametros de y pelo de x (some com os traços dos eixos)
plt.tick_params(axis='x', length=0)

#==========================================================================================
# Alterando cores das demais barras se não forem o valor máximo
for i, barra in enumerate (barras):
    if i != quantidades_ordenadas.index(max(quantidades_ordenadas)):
        barras[i].set_color(cores[1])
        barras[i].set_alpha(0.3)

#==========================================================================================
plt.title('Quantidade de Produtos Vendidos')

#==========================================================================================
# 2. comentar a linha de inversão de eixos
# 5. Descomentar a linha e inverter o eixo x
plt.gca().invert_xaxis()
plt.show()
```
