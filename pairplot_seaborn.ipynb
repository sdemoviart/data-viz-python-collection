# Pairplot

O Seaborn também possui uma função chamada pairplot, que cria um grid com scatterplot de todas as variáveis numéricas de um conjunto de dados. Além disso, os gráficos na diagonal principal são histogramas ou KDE das variáveis numéricas

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dados = 'churn_clientes.csv'

df_churn = pd.read_csv(dados)
```

```python
df_churn.head()
```

```python
sns.pairplot(data=df_churn, hue='churn')
```

```python
#plotando histogramas na diagonal
sns.pairplot(data=df_churn, hue='churn', diag_kind='hist');
```
